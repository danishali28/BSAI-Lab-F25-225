```python
%%writefile guided_chatbot.py
from lab6_search import Retriever
FAQ = [
    {"id": "faq:exams",
     "question": "Exam se aik raat pehle student itna serious kyun ho jata hai?",
     "answer": "Poora semester books nahi kholta, lekin exam se pehle sochta hai ke aik raat mein sab kuch yaad kar lega.",
     "source": "funny_faq.txt"},

    {"id": "faq:alarm",
     "question": "Student subah alarm lagane ke bawajood late kyun uthta hai?",
     "answer": "Raat ko 5 alarms lagata hai, lekin subah uthne ke bajaye sab ko band karke dobara so jata hai.",
     "source": "funny_faq.txt"},

    {"id": "faq:group_project",
     "question": "Group project mein sab se zyada kaam kaun karta hai?",
     "answer": "Aik banda poora project karta hai, doosra presentation deta hai aur baqi sirf apna naam likhwate hain.",
     "source": "funny_faq.txt"},

    {"id": "faq:salary",
     "question": "Student salary milne se pehle hi paisay kharch kyun kar deta hai?",
     "answer": "Paisay account mein aane se pehle hi shopping ki list ready hoti hai. Salary aati hai aur kuch din mein khatam ho jati hai.",
     "source": "funny_faq.txt"},

    {"id": "faq:programming",
     "question": "Programmer aik choti si coding mistake fix karne mein poora din kyun laga deta hai?",
     "answer": "Kabhi aik bracket ki wajah se poora code check karna parta hai. Itni mehnat ke baad pata chalta hai ke mistake sirf aik line mein thi.",
     "source": "funny_faq.txt"}
]
def build_bot(backend="tfidf"):
    records=[]
    for item in FAQ:
        data=dict(item)
        data["text"]=item["question"]+" "+item["answer"]
        records.append(data)
    return Retriever(records,backend)
def reply(engine,query):
    if not query.strip():
        return "Please enter a question."
    for item in FAQ:
        question =item["question"]
        if query.strip().casefold()   ==question.casefold():
            return item["answer"]+" [source: "+item["source"]+"]"
    hits=engine.search(query)
    if not hits:
        return "No matching FAQ. Please contact the help desk."
    lines=["Possible matches; choose or rephrase your question:"]
    for item,score in hits:
        question=item["question"]
        lines.append(question+f" (similarity {score:.3f})")
    return "\n".join(lines)
if __name__=="__main__":
    import argparse
    parser = argparse.ArgumentParser()
    parser.add_argument(
        "--backend",
        choices=["tfidf","embedding"],
        default="tfidf"
    )
    args=parser.parse_args()
    engine=build_bot(args.backend)
    print("FAQ assistant. Type exit to stop.")
    while True:
        try:
            query=input("You: ")
        except (EOFError, KeyboardInterrupt):
            break
        if query.strip().lower()=="exit":
            break
        answer=reply(engine, query)
        print("Bot:", answer)
```
